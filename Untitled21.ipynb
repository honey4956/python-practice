```python
try:
    num1=int(input("enter the first number:"))
    num2=int(input("enter the second numer:"))
    result=num1/num2
    print(f"the result is:{result}")
except zeroDivisionError:
    print("Error :you cannot divide a number by zero(0)")
except valueError:
    print("Error: please enter numbers only,no text")
finally:
    print("program excution completed.thank you")
    
    
             
```

```python
try:
    age=int(input("Enter your age"))
    if age < 0 or age > 150:
        print("Error:please enter a valid age enter a between 0 to 150.")
    else:
        print(f"your age is: {age}")
except valueError:
    print("Error:invalid age,please enter a valid age")
finally:
    print("validation check completed")
    
```
