```python
student={
    "name": "adeel",
    "age": 23,
    "city":"karachi"
}
student["email"]="adeel@example.com"

print(student)
```

```python
print(student["name"])
print(student["age"])
print(student["city"])
```

```python
print(student.get("name"))
print(student.get("age"))
print(student.get("city"))
```

```python
car={
    "brand" : "suzuki",
    "model" : "cultus",
    "year" : 2024
}
print(car.get("brand"))
print(car.get("model"))
print(car.get("year"))
car["color"] = "white"
print(car)
    
```

```python
car={
    "brand" : "suzuki",
    "year" : 2024,
    "model" : "cultus",
    "colors" : "white"
}
print(car.keys())
```

```python
print(car.values())
```

```python
print(car.items())
```

```python
employee={
    "name" : "adeel",
    "id" :102,
    "skills" :["python" , "SQL"],
    "project" :("project A","project B")
}
print(employee.get("skills"))
print(employee.get("name"))
print(employee.get("id"))
print(employee.get("project"))
employee["department"]="IT"
print(employee)
```
