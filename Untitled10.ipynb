```python
def count_vowels(text):
    vowels = "aeiouAEIOU"
    count = 0
    for char in text:
        if char in vowels:
            count = +1
    return count
result = count_vowels ('aeiouAEIOU')
print(f"output: {result} vowels")
```

```python
def hotel_bill(nights):
    def room_cost():
        return nights * 5000
    def tax():
        return room_cost() * 0.10
    def total_bill():
        return room_cost() + tax()

    print(f"room_cost:{room_cost()}")
    print(f"tax:{tax()}")
    print(f"total_bill:{total_bill()}")
hotel_bill(3)
```

```python
def shopping_bill(price,quantity,discount=0):
    def calculate_total():
        return price * quantity
    def apply_discount(total):
        final=total-discount
        if final < 0:
            return 0
        return final
    total_amount = calculate_total()
    final_bill = apply_discount(total_amount)
    return final_bill
    
print(f"final bill={shopping_bill(100,6)}PKR(no discount)")
print(f"final bill={shopping_bill(100,6, 100)}PKR(with discount)")
```
