```python
products = {
    "P1": {"name": "Shoes", "price": 2500, "stock": 5},
    "P2": {"name": "Shirt", "price": 1500, "stock": 10},
    "P3": {"name": "Bag", "price": 3000, "stock": 4},
    "P4": {"name": "Watch", "price": 4000, "stock": 3},
    "P5": {"name": "Headphones", "price": 2000, "stock": 6},
}
cart = {}
while True:
    print("\n" + "=" * 40)
    print("       ONLINE STORE MENU")
    print("=" * 40)
    print("1. View Products")
    print("2. Add to Cart")
    print("3. View Cart")
    print("4. Checkout")
    print("5. Exit")
    print("=" * 40)
    choice = input("Enter your choice (1-5): ").strip()
    if choice == "1":
        print("\n--- AVAILABLE PRODUCTS ---")
        print(f"{'ID':<6} {'Product':<15} {'Price (Rs.)':<15} {'Stock'}")
        print("-" * 45)
        for pid, details in products.items():
            print(
                f"{pid:<6} {details['name']:<15} {details['price']:<15} {details['stock']}"
            )
    elif choice == "2":
        pid = input("Enter Product ID (e.g., P1): ").strip().upper()

        if pid not in products:
            print("Invalid Product ID! Please select from P1 to P5.")
        else:
            qty = int(input(f"Enter quantity for {products[pid]['name']}: "))

            if qty <= 0:
                print("Quantity must be at least 1.")
            else:
                current_in_cart = cart.get(pid, 0)
                total_requested = current_in_cart + qty
                available_stock = products[pid]["stock"]

                if total_requested > available_stock:
                    print(
                        f"Sorry! Not enough stock. Available: {available_stock}, Already in cart: {current_in_cart}."
                    )
                else:
                    cart[pid] = total_requested
                    print(
                        f"Added! {products[pid]['name']} (Total in cart: {cart[pid]})"
                    )

    elif choice == "3":
        if not cart:
            print("\nYour cart is currently empty.")
        else:
            print("\n--- YOUR SHOPPING CART ---")
            print(f"{'Product':<15} {'Qty':<8} {'Price':<12} {'Subtotal'}")
            print("-" * 50)
            cart_total = 0

            for pid, qty in cart.items():
                name = products[pid]["name"]
                price = products[pid]["price"]
                subtotal = price * qty
                cart_total += subtotal
                print(f"{name:<15} {qty:<8} Rs. {price:<8} Rs. {subtotal}")

            print("-" * 50)
            print(f"Cart Total: Rs. {cart_total}")
    elif choice == "4":
        if not cart:
            print("\nYour cart is empty! Add items before checkout.")
        else:
            print("\n--- CHECKOUT BILL ---")
            cart_total = 0

            for pid, qty in cart.items():
                name = products[pid]["name"]
                price = products[pid]["price"]
                subtotal = price * qty
                cart_total += subtotal
                print(f"{name} x {qty} = Rs. {subtotal}")

            print("-" * 30)
            print(f"Subtotal: Rs. {cart_total}")
            if cart_total >= 5000:
                discount = cart_total * 0.10
                final_bill = cart_total - discount
                print(f"Discount (10%): -Rs. {discount:.0f}")
            else:
                discount = 0
                final_bill = cart_total
                print("Discount: Rs. 0 (Shop Rs. 5,000+ for 10% off)")

            print(f"Final Payable Amount: Rs. {final_bill:.0f}")
            print("-" * 30)

            for pid, qty in cart.items():
                products[pid]["stock"] -= qty

            cart.clear()
            print("Thank you for shopping! Stock updated and cart is now empty.")

    elif choice == "5":
        print("Exiting program. Goodbye!")
        break
    else:
        print("Invalid choice! Please enter a number between 1 and 5.")
```
