```python
class Employee:
    def __init__(self, name, id, address, salary, post, year_of_joining, education, year_of_experience):
        self.name = name
        self.id = id
        self.address = address
        self.salary = salary
        self.post = post
        self.year_of_joining = year_of_joining
        self.education = education 
        self.year_of_experience = year_of_experience
    def describe_employee(self):
        print(f"Name: {self.name}, ID: {self.id}, Post: {self.post}, Salary: {self.salary}, Experience: {self.year_of_experience} years")

    def update_year_of_experience(self, new_experience):
        self.year_of_experience = new_experience
        print(f"{self.name}'s experience has been updated to {self.year_of_experience} years.")
    def increase_salary(self, amount):
        self.salary += amount
        print(f"{self.name}'s new salary is {self.salary}.")
emp1 = Employee("adeel", 101, "Nazimabad", 40000, "Manager", 2020, "intermediate",6)
emp2 = Employee("hamza", 105, "liaqutabad town", 55000, "web develop", 2020, "intermdiate", 6)
emp3 = Employee("saim", 103, "baldia town", 30000, "stock exchange", 2022, "intermediate", 4)
emp4 = Employee("yasir", 107, "baldia town", 80000, "softwere house", 2021, "b.com", 5)
emp5 = Employee("saroosh", 108, "malir", 35000, "real state", 2023, "BSCS", 3)
employee_list = [emp1, emp2, emp3, emp4, emp5]

print("\n---All 5 employees ---")
for emp in employee_list:
    emp.describe_employee()
```
